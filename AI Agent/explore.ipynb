```python
import sqlite3
# Connect to the SQLite database
conn = sqlite3.connect('olist.sqlite')
# Create a cursor object
cursor = conn.cursor()
```

```python
# Execute a query
cursor.execute("""
SELECT
  distinct customer_state from customers  
  """)
# Fetch and print the results
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# Close the connection
conn.close()
```
